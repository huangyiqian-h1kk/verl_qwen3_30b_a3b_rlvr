# Qwen3-30B-A3B RLVR → Legal Reasoning Entropy Curve

This notebook reads the CSV files produced by `aggregate_legal_rlvr_entropy_curve_v1.py`. Set `OUT_ROOT` to the ABCI result directory.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

OUT_ROOT = Path('/groups/gcg51557/experiments/0390_rlsd/RLVR/verl_qwen3_30b_a3b_rlvr/outputs/legal_entropy_curve_qwen3_30b')
CURVES = OUT_ROOT / 'curves'
summary = pd.read_csv(CURVES / 'legal_entropy_curve_summary.csv')
summary
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
for hint, frame in summary.groupby('hint_level'):
    frame = frame.sort_values('checkpoint_step')
    axes[0].plot(frame.checkpoint_step, frame.full_entropy_mean, marker='o', label=hint)
    axes[1].plot(frame.checkpoint_step, frame.acc_source_law, marker='o', label=hint)
axes[0].set(title='Full policy entropy', xlabel='RLVR checkpoint step', ylabel='Entropy (nats)')
axes[1].set(title='Source-law accuracy', xlabel='RLVR checkpoint step', ylabel='Accuracy', ylim=(-0.02, 1.02))
for ax in axes:
    ax.grid(alpha=.25)
    ax.legend()
plt.tight_layout();
```

```python
keyword_summary = pd.read_csv(CURVES / 'keywords' / 'keyword_entropy_curve_summary.csv')
keyword_summary[['checkpoint_step', 'hint_level', 'keyword', 'phrase_entropy_macro_trajectory', 'keyword_trajectory_coverage']].head(20)
```
